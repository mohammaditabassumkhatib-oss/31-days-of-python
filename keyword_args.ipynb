`**kwargs` = Keyword arguments | Dictionary

```python
def add(**kwargs):
    print(type(kwargs)) # result = Dictionary
add()
```

```python
def customer_preference(**kwargs):
    print("Customer Preferences:")

    for i,j in kwargs.items():
        print(i,"=",j)

customer_preference(spicy=True,
                    cheese=False,
                    drink="Cola",
                    size="Large")
```

*API Requet parameters*

```python
def make_request(**kwargs):
    print("Request Parameters:")

    for key,value in kwargs.items():
        print(key,"=",value)

make_request(endpoint="/users",
             method = "GET",
             timeout = 30)
```

*LLM Configuration* - access a specific value with default (.get())

```python
def ask_llm(prompt,**settings):
    print("Prompt:",prompt)

    for key,value in settings.items():
        print(key,"=",value)

ask_llm("Explain neural networks",
    temperature=0.7,
    max_tokens=500,
    model="gpt")
```

*check whether a KEY exists*

```python
def tour(**kwargs):

    if "place" in kwargs:
        print(kwargs["place"],"has been selected for trip.")
    if "hotel" in kwargs:
        print(kwargs["hotel"], "is the destination.")

tour(place="Switzerland",hotel="InnKeeper")
```

*Get only the keys*

```python
def lifestyle(**kwargs):
    print("Wealthy lifestyle:")

    for key in kwargs:
        print(key)

lifestyle(Car=True,Land=True,Mansion=True, cycle = False)
```

*Get only values*

```python
def lifestyle(**kwargs):
    print("Wealthy lifestyle:")

    for value in kwargs.values():
        print(value)

lifestyle(Car="Rolls Royce",Land="100 Acre",Mansion="UK", cycle = "No")
```

*Modify a value*

```python
def family(**kwargs):

    if "family" in kwargs:
        kwargs["family"] = kwargs["family"].upper()

    return kwargs

family(family = "Rothschild")
```
