```python
numbers = [1, 2, 3, 4, 5]

# create cube as value in a dictionary
cube = {i:i**3 for i in numbers}
print(cube)
```

```python
words = ["Python", "AI", "Machine", "Learning"]

# key = words, value = length of word
length = {word:len(word) for word in words}
print(length)
```

```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]

# ONLY EVEN NUM
# key = num, value = square
even = {num: num**2 for num in numbers if num%2 == 0}
print(even)
```

```python
numbers = [1, 2, 3, 4, 5]

# even num as "EVEN" and Odd num as "ODD"
num_dict = {n:"EVEN" if n%2 == 0 else "ODD" for n in numbers}
print(num_dict)
```

```python
scores = {
    "Alice": 85,
    "Bob": 42,
    "Charlie": 91,
    "David": 67,
    "Eve": 78
}

# dictionary of students scored above 70
high_score = {name:score for name,score in scores.items() if score >=70}
print(high_score)
```
