`*args` = positional arguments, becomes tuple

```python
def argument_type(*args):
    print(type(args)) #result = tuple
argument_type()
```

```python
def add(*args):
    total = 0

    for number in args:
        total += number

    return total

print(add(10,20,30,40,50))
```

```python
# Function: It should multiply all the numbers passed to it.
def multiple(*args):
    total = 1

    for number in args:
        total*=number

    return total

multiple(2, 3, 4)
```

```python
# Function: It should print the person's name and then each skill.
def display_skills(name, *skills):
    print("Name:", name)

    for skill in skills:
        print(skill)

display_skills("Alex", "Python", "SQL", "Git")
```

```python
def add(a, b, c):
    return a + b + c

numbers = [10, 20, 30]

print(add(*numbers)) # (* = unpacks the list)
```

**Exercise: Restaurant Order System**

a function that calculates the total bill for any number of ordered items.
- Each item has a price.

Requirements:
1. Accept any number of prices using *args.
2. Add all the prices.
3. Return the total.
4. Don't use Python's built-in sum()
5. Bonus: the restaurant gives a 10% service charge.

```python
def calculate_bill(*prices):
    total = 0
    for price in prices:
        total+=price
    return total

calculate_bill(120, 250, 80, 150, 90)
```

```python
# Bonus part
def calculate_bill(*prices):
    total = 0
    for price in prices:
        total+=price
    service_charge = total*(1/10)
    total += service_charge
    return total

calculate_bill(120, 250, 80)
```

**Exercise: Restaurant Order by Customer**

The function should receive:
- customer_name → a normal parameter
- *prices → any number of food-item prices

Requirements inside the function:
1. Print the customer's name.
2. Loop through prices and calculate the food total.
3. Calculate a 10% service charge.
4. Calculate the final bill.
5. Print all three amounts.

```python
def customer_bill(name, *prices):
    print("Customer:", name)

    total = 0
    for price in prices:
        total += price
    print("Food total:",total)

    service_charge = total*0.10
    print("Service charge:",service_charge)

    total += service_charge
    print("Final bill:", total)

customer_bill("Alex", 120, 250, 80)
```

***Production-Level Exercise: Restaurant Order Processor***

Requirements inside a function must:
1. Accept customer_name as a normal parameter.
2. Accept any number of orders using *orders.
3. Unpack each order into item and price.
4. Print every valid order.
5. Reject orders where price <= 0.
6. Calculate the subtotal without using sum().
7. Add a 10% service charge.
8. Print the final total.
9. Handle this edge case: if there are no orders - program shouldn't crash.

```python
def process_order(customer_name, *orders):
    print("Customer:",customer_name)

    if orders == ():
        print("Make an order please!")
    else:
        print("Orders:")
        for order in orders:
            print("-",order)

        total = 0
        for  item,price in orders: # unpack the *orders
            if price <=0:
                return("Please enter a valid price!")
                
            else:
                total += price
        print("Subtotal:",total)

        service_charge = total*0.10
        print("Service charge:", service_charge)

        total += service_charge
        print("Final total:",total)


process_order(
    "Emma",
    ("Burger", 250),
    ("Invalid Item", -100),
    ("Pizza", 500)
)
process_order("John")
process_order(
    "Sarah",
    ("Pasta", 300),
    ("Salad", 150),
    ("Coffee", 120),
    ("Cake", 200),
    ("Water", 30)
)
```
